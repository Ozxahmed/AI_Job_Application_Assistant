```python
import asyncio
from dotenv import load_dotenv
import os
from langchain_openai import ChatOpenAI
from browser_use import Agent, Browser
```

```python
# load env variables
load_dotenv()

# initialize LLM engine
llm = ChatOpenAI(model="gpt-4o")

# define asyncio function
async def main():
    print("📡 Sending test ping to OpenAI...")

    # Send a fast test prompt to the model asynchronously
    response = await llm.ainvoke("Hello! Reply with 'Connection Successful' if you can read this.")

    # Print out the text content of the model's response
    print("\n🤖 Response from LLM:")
    print(response.content)
```

```python
# in python script file, use the following to call the event loop function:
# asyncio.run(main())

# In Jupyter, use 'await' directly at the top level
await main()
```

```python
from browser_use import Agent, Browser, ChatOpenAI
from dotenv import load_dotenv

load_dotenv()

llm = ChatOpenAI(model="gpt-4o")

browser = Browser(
    headless=False,
    executable_path="/Applications/Google Chrome.app/Contents/MacOS/Google Chrome",
    user_data_dir="./browser_profiles/chrome_agent",
    window_size={"width": 1280, "height": 900},
    keep_alive=True,
)

async def run_browser_test():
    task = "Go to https://wikipedia.org, search for 'Data Engineering', and tell me the first sentence of the page."

    agent = Agent(
        task=task,
        llm=llm,
        browser=browser,
    )

    print("🚀 Launching visible Vivaldi browser...")
    result = await agent.run()
    print(result)

await run_browser_test()
```
